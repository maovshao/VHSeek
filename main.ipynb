## iPHoP Benchmark — Summary and Distance Control


```python
import os
from logzero import logger
from vhseek.vhseek_util.statistics_util import major_test, major_plot, distance_control_plot

method_prediction_files_dict = {}
order_list = []

baseline_pred_dir = "vhseek_data/data/iphop/baseline"
# Ordered by current genus-level Fmax so all iPHoP major benchmark panels share the same x-axis.
for key in (
    "VirMatcher", "EvoMIL", "CRISPR", "SpacePharer", "PhisDetector",
    "PHP", "VPF-Class", "WIsH", "HostPhinder", "CHERRY",
    "VirHostMatcher-Net", "RaFAH", "Blast", "vHULK", "iPHoP"
):
    order_list.append(key)
    method_prediction_files_dict[key] = os.path.join(baseline_pred_dir, key)

model_files_dir = "vhseek_data/experiment/test_result/iphop/"
method_names_map = {
    'VHSeek (ESM2-650M)': "vhseek_esm2_650_probability",
    'VHSeek (AIDO-300M)': "vhseek_aido_300_probability",
    'VHSeek (Threshold)': "threshold/vhseek_probability",
    'VHSeek': "vhseek_probability",
}
for key in method_names_map:
    order_list.append(key)
    method_prediction_files_dict[key] = os.path.join(model_files_dir, method_names_map[key])

model_truth_paths_dict = {
    "virus_test_path": "vhseek_data/data/iphop/label_transfer/test",
    "label_taxonomy_path": "vhseek_data/data/util/host_label/label_transfer/taxonomy",
    "label_index_path": "vhseek_data/data/util/host_label/label_transfer/index",
}

baseline_truth_paths_dict = {
    "virus_test_path": "vhseek_data/data/iphop/baseline/label_transfer/virus_host_dict_test",
    "label_taxonomy_path": "vhseek_data/data/iphop/baseline/label_transfer/host_label_taxonomy",
    "label_index_path": "vhseek_data/data/iphop/baseline/label_transfer/host_label_index",
}

result_dict = {}
top_methods_order_list = order_list[-5:]

for key in order_list:
    pred_file = method_prediction_files_dict[key]
    if os.path.exists(pred_file):
        if key in method_names_map:
            result = major_test(pred_file, model_truth_paths_dict)
        else:
            result = major_test(pred_file, baseline_truth_paths_dict)
        if result:
            result_dict[key] = result
    else:
        logger.warning(f"Prediction file not found: {pred_file}. Skipping.")

major_plot(
    result_dict=result_dict,
    order_list=order_list,
    top_methods_order_list=top_methods_order_list,
    save_prefix="iPHoP_Major",
)

mash_dist_file = "vhseek_data/data/iphop/mash_similarity/iphop_test_distance_min"
thr_list = [0, 0.05, 0.10, 0.20, 0.40]

distance_control_plot(
    result_dict=result_dict,
    virus_distance_path=mash_dist_file,
    thresholds=thr_list,
    method_order=top_methods_order_list,
    full_order_list=order_list,
    level="genus",
    save_prefix="iPHoP_Distance_Control",
)
```
